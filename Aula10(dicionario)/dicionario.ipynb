# Dicionários

### Estrutura :

dicionario = {chave:valor,chave:valor......}

## vantagens e desvantagens

1. nao devem ser usados para pegar itens em uma determinada ordem
2. podem ter valores heterogêneos ( vários tipos de valores dentro de um mesmo dicionário: inteiros,strings,listas,etc)
3. Chaves são únicas obrigatoriamente
3. Mais intuitivos de trabalhar


```python
mais_vendidos = {'tecnologia': 'iphone', 'refrigeração': 'ar consul 12000bt', 'livros': 'o alquimista'}
vendas_tecnologia = {'iphone':1500,'samsung galaxy':12000,'tv samsung':10000,'ps5':14300,'tablet':1720,'ipad':1000}

```

* qual foi o item mais vendido nas categorias "livros"e "lazer"?
* qual foi o mais vendido de "notebook" e "ipad"?

```python

livro = mais_vendidos['livros']
qtde_ipad = vendas_tecnologia['ipad']

print(f"Livro mais vendido: {livro}")
print(f"Quantidade de iPads vendidos: {qtde_ipad}")
```

### duas formas de pegar um valor: 
* dicionário[chave]
* get(chave)

## o get() que aceita dois argumentos: *a chave do valor* a ser obtido e um *valor alternativo* a ser retornado se essa chave não existir.
### listaDeFrutas = {'macas': 5 ,'peras': 2}

```python
print(vendas_tecnologia.get('iphone'))
```
