```python
# Argumentos de ínicio,fim e incremento de range()
# O primeiro argumento, 12, é onde ele vai iniciar .
# O segundo argumento será o número com qual o loop deve terminar, ams sem concluir, logo ele para no 15;
for i in range(12,16):
     print(i)
```

```python
for i in range(0,10,2):
     print(i)
```

## A funçãp range() é flexível quanto à sequÊncia de números gerados em loops FOR.

estrutura:

digamos que estaja analisando a meta de vendas de vários funcionários de uma empresa. A meta de vendas é de 1000 reais em 1 dia.
Temos uma lista com as vendaas de todos os funcionários e quero calcular o % de pessoas que bateram a meta

```python
vendas = [1200,300,800,1500,1900,2750,400,20,23,70,90,80,1100,999,900,880,870,50,1111,120,300,450,800]
meta = 1000
quantidade_bateu_meta = 0
quantidade_de_funcionarios=len(vendas)
 
for venda in vendas: 
     if venda>=meta:
          print(venda)
          quantidade_bateu_meta +=1

     # print(venda)
print("A quantidade de pessoas que batem a meta {}".format(quantidade_bateu_meta))

quantidade_de_funcionarios=len(vendas)

print("O percentual de pessoas que bateram a meta foi de {}" .format(quantidade_bateu_meta/quantidade_de_funcionarios))

```

```python
#Exercicio 1
# Solicitar quantos números o usuário deseja informar, receber cada um deles e mostar a média aritmética no final.
soma = 0
num1 = int(input("Digite quantos números você deseja digitar: "))
for i in range(num1):
     num2 = int(input("Digite o número "))
     print("Número digitado ",num2) 
     soma = soma + num2
print("A soma dos número digitados é ",soma)
media = soma/num1
print("Média Aritmética é ", media)
 
```

```python
#Exercicio 2
# Receber 5 números e,ao final, informar quantos são positivos e quantos são negativos

num1 = int(input("Digite quantos números você deseja digitar: "))
negativo = 0
positivo = 0
nulo = 0
for i in range(num1):
      num2 = int(input("Digite o número "))
      if num2 < 0:
            negativo+=1
      elif num2 == 0:
          nulo+=1
      else:
          positivo+=1
print("Total de negativos foram ", negativo)
print("Total de positivos foram ", positivo)
print("Total de nulos foram ", nulo)


            
```
