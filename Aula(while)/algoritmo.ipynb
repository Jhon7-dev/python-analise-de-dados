# Estrutura while 

funcionamento: 
usando o while quando queremos repetir um código de forma indeterminada até uma condição se verdadeira
A lógica enquanto a condição for verdadeira, o while executa o código. Assim que ele termianar de ser verdadeiro, o código "sai" do while.



```python
while condicao:
     repete o codigo
```

```python
#exemplo
count = 0
while (count < 5): count += 1; print("Hello World")
```

```python
x = 0
while x < 10:
    x += 1
    if x == 5:
        continue
    print(x)
```

```python
#Exercicio 1 
# Solicitar quantos números o usuário deseja informar, receber cada um deles e mostrar a média aritmética no final
cont = 0
soma = 0
numero = int(input("Quantos números você deseja digitar? "))

while numero!=cont:

     num = int(input("digite o numero: " ))
     cont +=1
     soma  +=num 


media = (soma/numero)     
print("foram solicitados ao todo {} ".format(numero), "e a média aritmetica dele é {} ".format(media))
```

```python
#Receber 10 Números e, ao final informar quanos são positivos e quantos são negativos 

cont = 5
numeroPositivo = 0
numeroNegativo = 0
n = 1
while(n<cont):
     cont +=1
     num = int(input("Digite o número: "))
     if(num>0):
          numeroPositivo += 1
     elif(num<0):
          numeroNegativo += 1   
     else:
          print("ERROR!")
print("Ao total foram {}".format(numeroPositivo), " E um total de {}".format(numeroNegativo))
```
