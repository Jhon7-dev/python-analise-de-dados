"Fórmulas" de texto - métodos de String 

Estrutura: 
-> normalmente usamos a estrutura texto.metodo() para fazer as modificações que queremos 
-> alguns métodos peedem "argumentos", que são informações que temos que passar para a fórmula (método) para ela funcionar. Esses argumentos são passados dentro dos parênteses

Como usar:

capitalize() -> Coloca a 1ª letra Maiúscula

Uso:
  texto = 'joao'
  print(texto.capitalize())
Resultado:
     'Joao'

casefold() -> Transforma todas as letras em minúsculas (existe lower() mas o casefold é melhor)

Uso:
   texto = 'Joao'
   print(texto.casefold())
Resultado:
   'joao'
