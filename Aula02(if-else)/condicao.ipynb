```python
# criar programam para analisar o resultado de vendas de produtos da Amazon em um mês.
#Meta de Vendas do Iphone = 50.000 unidades
#quantidade vendida no mês = 65.300 unidades

# o seu progrma de avisar caso o produto tenha batido a meta do mês.
#caaso o produto tenha batido a meta do mês, exibir"Batemos a meta de iphone , vendemos "X" unidades"
#Se ele não bateu a meta do mês, o seu programa não deve fazer nada"
```

```python
meta = 50000
quantidadeDeVenda = 65300
if quantidadeDeVenda > meta:
     print("batemos a meta de Iphone, vendemos {} unidades ".format(quantidadeDeVenda)) #esse print ta dentro do TAB e faz parte do IF
print('fim do programa')
```
