## instruções def com parâmetros
 
a definição da funçaõ hello() nesse programa contém  um parâmetro chamado *name*. Um *parâmetro* é uma variável em que um argumento é armazenado quando ua função é chamada. Na primeira vez que a função *hello* é chamada, isso é feito com o argumento 'alice'. 
*obs*: o valor armazenado em um parâmetro é esquecido quando a função retorna. Exemplo se adicionarmos print(name) depois de hello(bob), um nameError será gerado pelo progrma, pois não há nenhuyma variável chamada *name*. 

```python
def hello(name):
     print('hello ' + name)
hello("joao")
hello("manu")
```

## valores de retorno e instruções return 
## o valor de uma instrução *return* é constituída das partes: 
* a palavra-chave return 
* o valor ou a espressão que a função deve retornar.

```python
import random 
def fazer_perguntas(number):
     if number == 1:
          return "certo!"
     elif number == 2:
          return "tente denovo!"
     elif number == 3:
          return 'yes'
     elif number == 4:
          return 'repita, tenta outra vez!'
     elif number == 5:
          return 'tente de novo!'
# r = random.randint(1,6)
# fortune = fazer_perguntas(r)
print(fazer_perguntas(random.randint(1,5)))     
```
