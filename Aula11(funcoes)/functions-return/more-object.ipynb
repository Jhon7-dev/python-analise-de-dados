```python
# como retornar mais de um objeto
# é possível retornar 2 "coisas",mais de uma coisa no pytohn
def operacoes_basicas(num1,num2):
     soma = num1 + num2
     diferenca = num1 - num2
     mult = num1 * num2
     divisao = num1/num2
     return soma,diferenca,mult,divisao
print(operacoes_basicas(100,20))
```
