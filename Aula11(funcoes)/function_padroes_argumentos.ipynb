```python
# vamos criar uma função que padronize códigos de produtos. O default será padronizar os códs para as letras minúsculas (dado por 'm'). mas se o usuário quiser pode padronizar para maiúscula, dado por ('M')

def padronizar_codigos(lista_codigos,padrao = 'm'):
     for item in lista_codigos:
          item = item.replace('  ','') 
          item = item.strip()
          if padrao == 'm':
               item = item.casefold()
          elif   padrao == 'M':
               item = item.upper()
          return lista_codigos
     return lista_codigos
cod_produtos = ['ABC12 ', 'abc34', 'AbC37']
print(padronizar_codigos(cod_produtos,'m'))
```
