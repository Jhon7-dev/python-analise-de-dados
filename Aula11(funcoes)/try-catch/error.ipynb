```python
def descobrir_servidor(email):
     try:
          # O bloco try except permite que você “tente” executar um bloco de código e, caso ocorra um erro, “capture” essa exceção e execute um código alternativo. 
          posicao_a = email.index('@')
          servidor = email[posicao_a:]
          if 'gmail' in servidor:
               return 'gmail'
          elif 'hotmail' in servidor or 'outlook' in servidor or 'live' in servidor:
               return 'hotmail'
          elif 'yahoo' in servidor:
               return 'yahoo'
          elif 'uol' in servidor or 'bol' in servidor:
               return 'uol'
     except:
          # Para mostrar detalhes específicos do erro ao usuário, podemos usar a sintaxe except Exception as. Isso nos permite capturar a exceção em uma variável e exibir informações detalhadas sobre o erro
          print('Email digitado não tem @, digite novamente')
          # O bloco else é executado apenas se o bloco try não gerar nenhuma exceção. Portan
     else:
          return 'não determinado'
     # O bloco finally é executado independentemente de qualquer exceção que ocorra. Isso é útil para liberar recursos ou executar ações de limpeza.
     finally:
         print('Execução finalizada')
email = input('qual o seu e-mail?')
print(descobrir_servidor(email))
```
