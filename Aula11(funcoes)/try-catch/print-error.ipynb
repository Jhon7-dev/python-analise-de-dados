```python
# passo1 => mostrar o erro de faturamento - custo se a pessoa digitar texto.
# passo2 => colocar um except para tratar o ValueError
# passo3 => usar um else para exibir o print (lucro)  e mostrar a difereça que seria de colocar print(lucro dentro do try)
# CASO o print (Lucro levantasse um valueERROR, ele ia para o excpet também), então muitas vezes é melhor isolar  => 
```

```python
custo = 500

try:
      faturamento = int(input('Qual foi o faturamento da loja no dia de hoje?'))
      lucro = int(faturamento) - int(custo)
      print(lucro)
except ValueError:
      raise Exception('Coloque apenas o valor do faturamento, sem texto nenhum, apenas números')
else:
     print(faturamento)
     
```
