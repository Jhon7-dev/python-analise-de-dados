```python
# function para carga tributária
# crie uma função que calcule qual o % de carga tributária que está sendo aplicado sobre um determinado produto, dado o preço de venda, o "lucro" e os custo (com exceção do imposto)dele.
preco = 1500
custo = 400
lucro = 800


def calculo_ct (preco_tex,custo_tex,lucro_tex):
     imposto = preco_tex - custo_tex - lucro_tex
     
     carga_tributaria = imposto / preco_tex
     return carga_tributaria

print('a carga tributária foi de {:} '.format(calculo_ct(preco,custo,lucro)))
```
