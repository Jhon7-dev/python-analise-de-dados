```python
# Uma empresa de pagamentos digitais deseja calcular sua taxa de fraude financeira. Essa taxa representa o percentual de dinheiro perdido em compras fraudulentas em relação ao total de compras válidas e fraudulentas.
# formula da fraude =  valor_das_compras_fraudulentas / (valor_das_compras_aprovadas + valor_das_compras_fraudulentas)

# Seu programa deve receber um dicionário contendo todas as transações da empresa. Cada chave do dicionário representa o código da transação e cada valor é uma tupla contendo:

```

```python
transacoes = {
    'TR001': (1500, 'Aprovada', ''),
    'TR002': (2300, 'Negada', 'Fraude'),
    'TR003': (800, 'Negada', 'Cartão sem limite'),
    'TR004': (4200, 'Aprovada', ''),
    'TR005': (3100, 'Negada', 'Fraude'),
    'TR006': (950, 'Aprovada', ''),
    'TR007': (1200, 'Negada', 'Dados inválidos'),
    'TR008': (2800, 'Aprovada', ''),
    'TR009': (1750, 'Negada', 'Fraude'),
    'TR010': (5100, 'Aprovada', ''),
    'TR011': (990, 'Negada', 'Cartão sem limite'),
    'TR012': (6200, 'Aprovada', ''),
    'TR013': (4100, 'Negada', 'Fraude'),
    'TR014': (1300, 'Aprovada', ''),
    'TR015': (2200, 'Negada', 'Dados inválidos'),
}
def taxaFraude(arquivos):
     denominador = 0
     numerador = 0
     for arquivo in arquivos:
          valor,status,motivo = arquivos[arquivo]
          if status == 'Aprovada':
               denominador+=valor
          elif status == 'Negada' and motivo ==  'Fraude':
               numerador+=valor
               denominador+=valor
     return numerador / denominador
print('{:.2%}'.format(taxaFraude(transacoes)))

               
```
