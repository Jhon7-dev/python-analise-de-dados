Exercício 1 — Sistema de Produtos Congelados

Uma distribuidora possui diversos produtos armazenados em estoque.
Cada produto possui um código e as 3 primeiras letras indicam sua categoria.

Categorias:
CON → Produtos congelados
FRT → Frutas
LIM → Produtos de limpeza
Exemplos:
Pizza congelada → CON45871
Lasanha congelada → CON98541
Maçã → FRT15478
Detergente → LIM78451

Seu objetivo é criar uma função que:

Receba um código de produto.
Verifique se ele pertence ao setor de congelados.
Retorne True ou False.

Depois:

Percorra a lista de produtos.
Mostre uma mensagem para os produtos congelados dizendo que devem ser enviados para a câmara fria.


```python
def congelados(produto_congelado):
      produto_congelado = produto_congelado.upper()
      if 'CON' in produto_congelado:
           return True
      else:
           return False

produtos =[
    'CON45871',
    'FRT15478',
    'LIM78451',
    'CON98541',
    'FRT44785',
    'CON33669',
    'LIM99874',
    'CON74125',
    'FRT23569',
    'CON55214'
]
for produto in produtos:
     if congelados(produto):
          print("Enviando para a câmara fria  " + produto) 
     else:
          print('Enviando para outro setor')         
```

Exercício 2 — Controle de Chamados de TI

Uma empresa possui um sistema de chamados técnicos.
Cada chamado possui um código e as 3 primeiras letras indicam o setor responsável.

Categorias:
RED → Problemas de rede
HAD → Problemas de hardware
SFT → Problemas de software
Exemplos:
Internet caiu → RED45871
Computador não liga → HAD99874
Erro no sistema → SFT55214

Crie um programa que:

Faça uma função chamada problema_rede()
A função deve verificar se o chamado pertence ao setor de rede.
Retorne True ou False.

Depois:

Percorra a lista de chamados.
Exiba uma mensagem direcionando os chamados de rede para a equipe especializada.

startswith() é um método do Python usado para verificar se uma string COMEÇA com determinado texto

```python
def problema_rede(rede):
     rede = rede.upper()
     if "RED" in rede:
          return True
     else:
          False
chamados = [
    'RED45871',
    'HAD15478',
    'SFT78451',
    'RED98541',
    'SFT44785',
    'RED33669',
    'HAD99874',
    'RED74125',
    'SFT23569',
    'RED55214'
]

for chamado in chamados:
     if problema_rede(chamado):
          print('problemas de rede')
          print(f'Enviar {chamado} para o setor de redes ')
     elif chamado.startswith('HAD'):
          print(f'Enviar {chamado} para setor de hardware')
     elif chamado.startswith('SFT'):
          print(f'Enviar {chamado} para setor de software')
```
