# Casos com varias condições/comparações

Estrutura:

quando temos várias comparações, ao invés de criar "if" dentro de "if" podemos usar os operadores "and"e"or" para tratar essas condições.
Funciona assim: 

 if condicao_1 and condicao_2:
      
vai ser executado se as 2 condições forem verdadeiras ao mesmo tempo
outro caso:

 if condicao_1 or condicao_2:
     
vai ser executado se pelo menos uma das condições forem verdadeiras

```python
#Exemplo 
# Se o funcionário vendeu mais do que a metada de vendas e a loja bateu a meta de vendas da loja, o funcionário ganha 3% do vendeu em forma de bônus.
# Caso o funcionário tenha batido a meta de vendas individual dele, mas a loja não tenha batido a met de vendas da loja como um todo, o funcionário não ganha bônus.
```

```python
meta = 40.000
metaDaLoja = 50.000
metaDaLojaBatida = 60.000
venda = 34.000

if(venda > meta/2 and metaDaLojaBatida>metaDaLoja):
     bonus = venda*0.03
     print("a meta foi batida! Ganhou {} de bônus".format(bonus))
if(venda>meta and metaDaLoja):
     print("Sem ganho de bônus")
```
