Listas em python

Estrutura: 

lista = [valor,valor,valor,valor...]

-> Listas em python foram feitas para serem homogêneas, apesar de aceitarem valores heterogêneos

exemplo: 
Listas de Produtos de uma loja: 


```python
produtos = ['tv','celular','mouse','teclado','tablet']

#Lista de Unidades de cada produto da loja

vendas =[1000,1500,350,270,900]

print(produtos)
print(vendas)
```

Listas em Python

Estrutura: 


lista = [valor,valor,valor,valor....]

lista[i] -> é o valor do índice "i" da lista
obs: o indice começa em zero, então o primeiro item de uma lista é o item lista[0]
Paa substituir um valor de uma lista você pode fazer: 

lista[i] = novo_valor
