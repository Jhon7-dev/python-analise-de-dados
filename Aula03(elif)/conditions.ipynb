```python
# ==== EXEMPLO =====
# criar um progrmama para analisar o bônus dos funcionparios de uma empresa.
# para os cargos de vendedores, a regra do bônus é de acordo com a meta de vendas da pessoa: 

# Se ela vendeu abaixo da meta dela, ela não ganha bônus.

# Se ela vendeu acima da meta dela, ela ganha como bônus 3% do valor que ela vendeu.

# Se ela vendeu mais do que o dobro da meta, ela ganha como bônus 7% do valor que ela vendeu.

# vamos criar um  progrmama para avaliar uma pessoa que tinha como meta de vendas 20.000 reais e calcular o bônus dela de acordo com o valor
```

```python
meta = 20000

venda = int(input("total de vendas: "))
if venda>=meta:
     bonus = 0.03 * venda
     print('parabéns, você bateu a meta! Ganhou {} de bônus'.format(bonus))
      
elif venda>2*meta:
     bonus = 0.07 * venda
     print('parabéns, você bateu o dobro meta! Ganhou {} de bônus  '.format(bonus))
      
else:
     print('não ganhou bônus! a meta era '.format(meta))
     
```
